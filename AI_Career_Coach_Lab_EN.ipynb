### 📚 Core Concepts to Know
Before we start coding, let's define our core foundation:
1. **LLM (Large Language Model):** The intelligence engine (such as Gemini). Think of it as an extremely advanced auto-complete engine trained on a vast corpus of human knowledge. It understands patterns, context, and semantic meaning.
2. **Prompt Engineering:** The art and science of communicating with LLMs. It is the process of structuring inputs to guide the model toward the most accurate, relevant, and useful outputs.
3. **AI Engineering:** The discipline of integrating an AI model into a real-world software system. While prompt engineering focuses on the *text*, AI engineering focuses on the *pipeline* (handling API keys, enforcing structured JSON output, managing state, error handling, and scaling).

## 👋 Section 3: Hello Gemini

Let's perform a quick connection test to ensure everything is working. We will initialize the `gemini-2.5-flash` model, optimized for fast, versatile, and highly efficient tasks.

```python
# Initialize the model instance
model = genai.GenerativeModel('gemini-2.5-flash')

# Send our first prompt
response = model.generate_content("Hello! I am ready to learn AI Engineering. Give me a 1-sentence welcome message.")

print("Gemini says:", response.text)
```

## 📄 Section 4: Input Data (Our User's Resume)

In AI Engineering, we dynamically inject user data into our prompts. Let's define the resume we are working with today.
We will store this in a variable so we can easily change it later without rewriting all our prompts.

```python
my_resume = """
Experience: 3 years in a retail call center. Handled customer complaints, managed the ticketing queue using Zendesk, and de-escalated angry clients. Promoted to Shift Lead in my final year.
Skills: Active listening, conflict resolution, basic HTML/CSS (self-taught), scheduling.
Education: Associate's Degree in General Business.
"""

print("Resume loaded into memory.")
```

## 🎯 Section 5: Zero-Shot Prompting

**Zero-Shot** means asking the AI a question directly without providing any examples of what the correct output should look like. Let's see how Gemini handles our resume out-of-the-box.

```python
prompt = f"""
Based on this resume, what are 3 good tech-industry career transitions for this person?

Resume:
{my_resume}
"""

response = model.generate_content(prompt)
print(response.text)
```

## 🎭 Section 6: Persona Prompting (Role-Playing)

The output above is decent, but it can sound a bit generic or robotic. We can dramatically alter the tone and expertise of the LLM by assigning it a persona. Let's ask it to act as an elite Silicon Valley technical recruiter.

```python
persona_prompt = f"""
Act as an elite Silicon Valley technical recruiter and empathetic career coach.
Review the following resume and suggest 3 modern tech roles they could transition into.
Use an encouraging, highly professional tone and use modern tech industry terminology.

Resume:
{my_resume}
"""

response = model.generate_content(persona_prompt)
print(response.text)
```

## 📝 Section 7: Few-Shot Prompting

**Few-Shot Prompting** involves providing the model with one or more examples of the input-output pairs you are looking for.
This is incredibly powerful for teaching the model a specific formatting pattern or logical flow without requiring verbose explanations.

```python
few_shot_prompt = f"""
Your task is to suggest career paths based on a user's resume, outputting strictly in the requested format.

Example 1:
USER RESUME: 5 years retail worker at a clothing store.
CAREER MATCH 1: Sales Development Representative (SDR)
CAREER MATCH 2: Customer Success Manager (CSM)

Example 2:
USER RESUME: High school math teacher.
CAREER MATCH 1: Data Analyst
CAREER MATCH 2: Instructional Designer

Now do it for this new user:
USER RESUME:
{my_resume}
"""

response = model.generate_content(few_shot_prompt)
print(response.text)
```

## ⚙️ Section 8: System Instructions

Instead of stuffing the persona into every single prompt, we can set **System Instructions**.
System instructions reside outside the standard prompt context and act as a permanent "prime directive" for the model instance.

```python
# We initialize a new model instance explicitly with a system instruction
coach_model = genai.GenerativeModel(
    'gemini-2.5-flash',
    system_instruction="You are a tough-love, highly pragmatic career coach. You prefer giving practical, direct advice without fluff. Always output bullet points."
)

response = coach_model.generate_content(f"What roles fit my resume? {my_resume}")
print(response.text)
```

## 🧠 Section 9: Chain of Thought (CoT)

Large Language Models (LLMs) reason significantly better when given "time to think". If we force the model to provide a final answer immediately, it might hallucinate or miss critical nuances.
**Chain of Thought (CoT) Prompting** means guiding the model to break down its reasoning process step-by-step *before* reaching the final conclusion.

### Chain of Thought Prompt Structure:

A Chain of Thought prompt typically consists of the following components:

1. **Initial Directive:** A clear instruction defining the overall task.
2. **Input Data:** The context or data the model needs to analyze (in our case, the resume).
3. **Step-by-Step Breakdown:** A list of sequential reasoning steps the model must follow before producing the final result. For example:
    * "Step 1: Extract X information from the input data."
    * "Step 2: Analyze Y based on the extracted X."
    * "Step 3: Synthesize results from Step 1 and Step 2 to generate Z."

Decomposing prompts into sequential steps helps the model reason methodically, drastically boosting output quality and accuracy—much like asking a student to show their work when solving a complex problem.

```python
cot_prompt = f"""
Review this resume:
{my_resume}

Follow these steps precisely:
Step 1: Extrapolate 3 abstract cognitive skills this person uses derived from their day-to-day tasks.
Step 2: Identify 2 tech-industry problems this person would be uniquely good at solving based on those skills.
Step 3: Finally, suggest 2 exact job titles that solve those problems, and provide a 1-sentence justification for each.
"""

# Let's use our base model for this one
response = model.generate_content(cot_prompt)
print(response.text)
```

## 🧱 Section 10: Structured Output (JSON) - The Core of AI Engineering

In a real-world web application, you don't want an unstructured wall of markdown text from an API; you need structured data that can be parsed and rendered by a frontend UI (such as JSON).
Here, we transition from Prompt Engineering to true **AI Engineering** by utilizing Gemini's native `response_schema`.

```python
import json
import typing_extensions as typing

# We use Python TypedDicts to natively define the exact JSON structure we want.
class CareerSuggestion(typing.TypedDict):
    job_title: str
    match_percentage_confidence: int
    missing_skills_to_learn: list[str]
    why_it_fits: str

# We pass this into GenerationConfig
response = model.generate_content(
    f"Analyze this resume and provide 2 tech career recommendations: {my_resume}",
    generation_config=genai.GenerationConfig(
        response_mime_type="application/json",
        response_schema=list[CareerSuggestion]
    )
)

# Parse the text response into an actual Python dictionary to prove it works
structured_data = json.loads(response.text)
print(json.dumps(structured_data, indent=2))
```

## 💬 Section 11: Multi-turn Conversations (Chat API)

Career coaching isn't a one-way transaction—it's an interactive dialogue.
To build a chatbot, we don't call `generate_content()` in isolation (which has no memory). Instead, we use `start_chat()`, which automatically tracks conversation history and context across multiple turns!

```python
# Start a fresh chat session using our "coach model" with the system personality
chat_session = coach_model.start_chat()

print("--- Turn 1 ---")
msg1 = f"Hi coach, I'm thinking of moving to tech. Here is my resume:\n{my_resume}\nWhat is 1 role I should pursue?"
res1 = chat_session.send_message(msg1)
print(res1.text)

print("\n--- Turn 2 (Testing Memory) ---")
# Notice we don't need to re-send the resume! The chat session remembers it.
res2 = chat_session.send_message("What is the very first technical skill I should learn for that specific role?")
print(res2.text)
```
